```python
import pandas as pd
import numpy as np


# ============================================================
# 1. FUNCTION TO CALCULATE GRADE
# ============================================================

def calculate_grade(average):
    
    if average >= 90:
        return "A+"
    
    elif average >= 80:
        return "A"
    
    elif average >= 70:
        return "B"
    
    elif average >= 60:
        return "C"
    
    elif average >= 50:
        return "D"
    
    else:
        return "F"


# ============================================================
# 2. FUNCTION TO CALCULATE RESULT
# ============================================================

def calculate_result(average, attendance):
    
    if average >= 40 and attendance >= 75:
        return "Pass"
    
    else:
        return "Fail"


# ============================================================
# 3. READ DATASET
# ============================================================

df = pd.read_csv("student.csv")


# ============================================================
# 4. DISPLAY STUDENT DATA
# ============================================================

print()
print("============================================================")
print("              STUDENT PERFORMANCE ANALYTICS SYSTEM")
print("============================================================")

print()
print("************ STUDENT DATA ************")
print()

print(df)


# ============================================================
# 5. SUBJECT COLUMNS
# ============================================================

subjects = ["Python", "DBMS", "CN", "Maths"]


# ============================================================
# 6. CALCULATE TOTAL MARKS
# ============================================================

df["Total"] = df[subjects].sum(axis=1)


# ============================================================
# 7. CALCULATE AVERAGE
# ============================================================

df["Average"] = df[subjects].mean(axis=1)


# ============================================================
# 8. CALCULATE GRADE
# ============================================================

df["Grade"] = df["Average"].apply(calculate_grade)


# ============================================================
# 9. CALCULATE RESULT
# ============================================================

df["Result"] = df.apply(
    lambda row: calculate_result(
        row["Average"],
        row["Attendance"]
    ),
    axis=1
)


# ============================================================
# 10. DISPLAY STUDENT PERFORMANCE
# ============================================================

print()
print("************ STUDENT PERFORMANCE ************")
print()

print(
    df[
        [
            "Student ID",
            "Name",
            "Total",
            "Average",
            "Grade",
            "Result"
        ]
    ]
)


print()
print("=" * 60)


# ============================================================
# 11. OVERALL CLASS AVERAGE
# ============================================================

class_average = np.mean(df["Average"])


print()

print(
    "Overall Class Average:",
    round(class_average, 2)
)


print()
print("=" * 60)


# ============================================================
# 12. HIGHEST PERFORMER
# ============================================================

top_student = df.loc[df["Average"].idxmax()]

print()
print("************ TOP PERFORMER ************")
print()

print("Name:", top_student["Name"])
print("Average:", round(top_student["Average"], 2))


print()
print("=" * 60)


# ============================================================
# 13. LOWEST PERFORMER
# ============================================================

lowest_student = df.loc[df["Average"].idxmin()]

print()
print("************ LOWEST PERFORMER ************")
print()

print("Name:", lowest_student["Name"])
print("Average:", round(lowest_student["Average"], 2))


print()
print("=" * 60)


# ============================================================
# 14. PASS / FAIL STATISTICS
# ============================================================

pass_count = (df["Result"] == "Pass").sum()

fail_count = (df["Result"] == "Fail").sum()

print()
print("************ PASS / FAIL STATISTICS ************")
print()

print("Passed Students:", pass_count)
print("Failed Students:", fail_count)


print()
print("=" * 60)


# ============================================================
# 15. SUBJECT-WISE AVERAGE
# ============================================================

print()
print("************ SUBJECT-WISE AVERAGE ************")
print()

for subject in subjects:
    
    average = np.mean(df[subject])
    
    print(
        subject,
        ":",
        round(average, 2)
    )


print()
print("=" * 60)


# ============================================================
# 16. TOP 5 STUDENTS
# ============================================================

print()
print("************ TOP 5 STUDENTS ************")
print()

top5 = df.sort_values(
    by="Average",
    ascending=False
).head(5)


print(
    top5[
        [
            "Name",
            "Average",
            "Grade",
            "Result"
        ]
    ]
)


print()
print("=" * 60)

```
